```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import  OneHotEncoder,OrdinalEncoder,LabelEncoder
from sklearn.impute import SimpleImputer
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.drop('has_covid',axis =1),df.iloc[:,-1],test_size=0.2, random_state=0)
```

```python
X_train.shape
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OneHotEncoder(sparse_output=False,drop='first'),['gender','city']),
    ('tnf3',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),

],remainder='passthrough')
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test)
```
